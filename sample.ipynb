```python
# Yahoo!ニュースのトピック一覧ページからタイトルとカテゴリを取得

import requests
from bs4 import BeautifulSoup
import pandas as pd

def get_titles_and_categories(url):
    # URLにアクセスしてページのHTMLを取得
    response = requests.get(url)
    soup = BeautifulSoup(response.content, 'html.parser')
    
    # 各カテゴリーとその記事タイトルを抽出
    sections = soup.find_all('div', class_='sc-hwj2au-1')
    data = []

    for section in sections:
        category = section.find('a').get_text(strip=True)
        articles = section.find_all('li', class_='sc-1nhdoj2-0')
        for article in articles:
            title = article.find('a').get_text(strip=True)
            data.append({"Category": category, "Title": title})
    
    return data

# Yahoo!ニュースのトピック一覧ページからタイトルとカテゴリを取得
url = "https://news.yahoo.co.jp/topics"
data = get_titles_and_categories(url)

# データをデータフレームに変換
df = pd.DataFrame(data)

# データフレームをCSVファイルに保存
df.to_csv('news_titles_and_categories.csv', index=False, encoding='utf-8-sig')

print("カテゴリと記事タイトルがnews_titles_and_categories.csvに保存されました。")
```

```python
# ニュースサイトから、一ヶ月以内の記事を取得する

import requests
from bs4 import BeautifulSoup
import pandas as pd
from datetime import datetime, timedelta

# 基本URLを設定します
base_url = "https://www.kyoceradocumentsolutions.com"

# ニュースのURLを設定します
news_url = "https://www.kyoceradocumentsolutions.com/ja/news/rls_2024/"

# ページを取得します
response = requests.get(news_url)
soup = BeautifulSoup(response.text, "html.parser")

# 現在の日付を取得します
now = datetime.now()
one_month_ago = now - timedelta(days=40)

# データを格納するリストを準備します
data = []

# ニュースエントリーをすべて抽出します
entries = soup.find_all("li")
for entry in entries:
    date_div = entry.find("div", class_="news-entries__date")
    category_div = entry.find("div", class_="news-entries__category")
    title_div = entry.find("div", class_="news-entries__title")
    
    # 必要な要素がすべて存在するか確認します
    if date_div and category_div and title_div:
        date_str = date_div.text.strip()
        category = category_div.text.strip()
        title = title_div.text.strip()
        relative_link = title_div.find("a")["href"]
        full_link = base_url + relative_link  # フルパスに変換
        
        # 日付をdatetimeオブジェクトに変換します
        entry_date = datetime.strptime(date_str, "%Y/%m/%d")
        
        # カテゴリが「製品・ソリューション」であり、かつ一ヶ月以内に更新された記事のみを抽出します
        if category == "製品・ソリューション" and entry_date >= one_month_ago:
            data.append([entry_date, category, title, full_link])

# pandasのDataFrameを作成します
df = pd.DataFrame(data, columns=["日付", "カテゴリ", "タイトル", "リンク"])

# DataFrameをcsvファイルに出力します
df.to_csv("products_solutions_updates.csv", index=False, encoding="utf-8")

print("CSVファイルに出力しました")
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
from datetime import datetime, timedelta

# 現在の日付を取得します
current_date = datetime.now()

# サイトのURL
url = "https://newsroom.lexmark.com/newsreleases"

# リクエストを送信してHTMLを取得します
response = requests.get(url)
soup = BeautifulSoup(response.content, "html.parser")

# ニュース記事を取得します
articles = soup.find_all("li", class_="wd_item")

# データを格納するリスト
data = []

# 各記事を解析します
for article in articles:
    date_str = article.find("div", class_="wd_date").text.strip()
    date = datetime.strptime(date_str, "%b %d, %Y")
    title = article.find("div", class_="wd_title").text.strip()
    summary = article.find("div", class_="wd_summary").text.strip()
    
    # 更新日が一ヶ月以内かどうかを確認します
    if (current_date - date) <= timedelta(days=30):
        data.append({
            "更新日": date_str,
            "タイトル": title,
            "概要": summary
        })

# データをDataFrameに変換します
df = pd.DataFrame(data)

# CSVファイルに保存します
df.to_csv("recent_articles.csv", index=False)

print("CSVファイルが生成されました。")
```

```python
以下の #要件 を満たすPython のコードを作成してください。
#前程条件：
・特定のフォルダ内に複数のcsvファイルがある

#要件：
・ライブラリは、tkinterを使用する
・任意のキーワードを入力すると、そのキーワードが文字内部に含まれるcsvファイルを検索する
・検索フォルダは、pythonファイルと同じフォルダ内とする
・検索する文字列は、部分一致でも可として、大文字と小文字は区別しない
・複数の検索結果があるときには、すべて表示する
・検索された「csvファイル名」と「検索されたセル内の文字列全体」を表形式で表示する
・表示されたファイル名をクリックした場合、csvファイルが開くことができる
・検索のステータスバーを表示する
・GUIのデフォルトサイズは800×800とする
```

```python
import os
import csv
import tkinter as tk
from tkinter import filedialog, messagebox
from tkinter import ttk

# キーワードでCSVファイルを検索する関数
def search_csv_files(keyword):
    results = []
    keyword = keyword.lower()
    for file in os.listdir('.'):
        if file.endswith('.csv'):
            with open(file, newline='', encoding='utf-8') as csvfile:
                reader = csv.reader(csvfile)
                for row in reader:
                    for cell in row:
                        if keyword in cell.lower():
                            results.append((file, cell))
                            break
    return results

# 検索ボタンがクリックされたときに呼び出される関数
def on_search():
    keyword = keyword_entry.get()
    if not keyword:
        messagebox.showwarning("入力エラー", "キーワードを入力してください")
        return

    status_label.config(text="検索中...")
    root.update_idletasks()

    results = search_csv_files(keyword)
    for row in tree.get_children():
        tree.delete(row)

    if results:
        for file, cell in results:
            tree.insert("", "end", values=(file, cell))
    else:
        messagebox.showinfo("結果なし", "該当するCSVファイルが見つかりませんでした")

    status_label.config(text="検索完了")

# CSVファイルを開く関数
def open_csv_file(event):
    item = tree.identify('item', event.x, event.y)
    if not item:
        return

    file = tree.item(item, 'values')[0]
    os.startfile(file)

# GUIのセットアップ
root = tk.Tk()
root.title("CSVファイル検索")
root.geometry("800x800")

frame = ttk.Frame(root)
frame.pack(padx=10, pady=10, fill="x", expand=True)

keyword_label = ttk.Label(frame, text="キーワード:")
keyword_label.pack(side="left", padx=(0, 10))

keyword_entry = ttk.Entry(frame)
keyword_entry.pack(side="left", fill="x", expand=True)

search_button = ttk.Button(frame, text="検索", command=on_search)
search_button.pack(side="left", padx=(10, 0))

tree = ttk.Treeview(root, columns=("ファイル名", "内容"), show="headings")
tree.heading("ファイル名", text="ファイル名")
tree.heading("内容", text="内容")
tree.pack(padx=10, pady=10, fill="both", expand=True)

tree.bind("<Double-1>", open_csv_file)

status_label = ttk.Label(root, text="ステータスバー")
status_label.pack(side="bottom", fill="x")

root.mainloop()
```

```python
import spacy
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import VotingClassifier
from sklearn.pipeline import make_pipeline

# spaCyの英語と日本語モデルをロード
nlp_en = spacy.load("en_core_web_sm")
nlp_ja = spacy.load("ja_core_news_sm")

# ニュースのタイトルとカテゴリのサンプルデータ
titles = [
    "最新のスマートウォッチが市場に投入されました",
    "企業の業績が大幅に向上しました",
    "革新的なタブレットが登場",
    "新しいCEOが就任しました",
    "次世代のゲームコンソールが発売されました",
    "会社の株価が急上昇しました",
    "新しいノートパソコンが発表されました",
    "企業の新しい戦略が公開されました",
    "新しいスマートフォンアプリがリリースされました",
    "取締役会のメンバーが変更されました",
    "最新のワイヤレスイヤホンが販売開始されました",
    "大規模なリストラが発表されました",
    "新しい自動車モデルが登場しました",
    "会社の新しい本社ビルが完成しました",
    "新しいスマートホームデバイスが市場に投入されました",
    "企業の年間収益が記録的な数字に達しました"
]
categories = [
    "新製品情報", "会社の重要情報", "新製品情報", "会社の重要情報",
    "新製品情報", "会社の重要情報", "新製品情報", "会社の重要情報",
    "新製品情報", "会社の重要情報", "新製品情報", "会社の重要情報",
    "新製品情報", "会社の重要情報", "新製品情報", "会社の重要情報"
]

# テキストの前処理関数
def preprocess_text(text):
    doc_ja = nlp_ja(text)
    doc_en = nlp_en(text)
    tokens_ja = [token.lemma_ for token in doc_ja if not token.is_stop]
    tokens_en = [token.lemma_ for token in doc_en if not token.is_stop]
    return " ".join(tokens_ja + tokens_en)

# 前処理済みのテキストを取得
preprocessed_titles = [preprocess_text(title) for title in titles]

# ロジスティック回帰とランダムフォレストを使ったアンサンブルモデルの作成
model = make_pipeline(
    TfidfVectorizer(),
    VotingClassifier(
        estimators=[
            ('lr', LogisticRegression(max_iter=1000)),
            ('rf', LogisticRegression(max_iter=1000))
        ],
        voting='soft'
    )
)

# モデルの訓練
model.fit(preprocessed_titles, categories)

# 新しいニュースタイトルの分類
new_titles = [
    "最新のタブレットが登場しました",
    "会社の新しいプロジェクトが始動しました",
    "次世代のスマートフォンが発表されました",
    "新しいCOOが選任されました",
    "音声をリアルタイムで表示する字幕表示システム「CotopatⓇ Mobile」バージョンアップ版を京セラドキュメントソリューションズジャパンより販売開始",
    "カラーA4プリンターECOSYS PA2600cwx、カラーA4複合機ECOSYS MA2600cwfxを新発売",
    "京セラ株式会社 2025年3月期 第3四半期 決算発表"
]
preprocessed_new_titles = [preprocess_text(title) for title in new_titles]
predicted_categories = model.predict(preprocessed_new_titles)

# 分類結果の表示
for title, category in zip(new_titles, predicted_categories):
    print(f"タイトル: {title} => カテゴリ: {category}")
```

```python
import requests
from bs4 import BeautifulSoup
from gensim.summarization import summarize

# URLリスト
urls = ["https://news.yahoo.co.jp/pickup/6544667"]

# 保存するファイル名
output_file = "summary.txt"

with open(output_file, "w", encoding="utf-8") as file:
    for url in urls:
        response = requests.get(url)
        
        if response.status_code == 200:
            soup = BeautifulSoup(response.text, 'html.parser')
            
            # ページの本文を抽出 (例: <p>タグ)
            paragraphs = soup.find_all('p')
            content = " ".join([p.get_text(strip=True) for p in paragraphs])
            
            # 要約を生成
            if content:
                try:
                    summary = summarize(content, word_count=50)  # 要約の単語数を調整
                except ValueError:
                    summary = "要約が生成できませんでした。本文が短すぎるかもしれません。"
            else:
                summary = "本文が見つかりませんでした。"
            
            # 結果を保存
            file.write(f"URL: {url}\n")
            file.write(f"要約:\n{summary}\n\n")
        else:
            file.write(f"URL: {url} の取得に失敗しました。\n\n")

print(f"要約を{output_file}に保存しました！")
```

```python
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from bs4 import BeautifulSoup
import pandas as pd
import time

url = "https://www.hp.com/us-en/shop/pdp/hp-color-laserjet-enterprise-6700dn-printer#techSpecs"

options = Options()
options.add_argument("--headless")
driver = webdriver.Chrome(options=options)

print("Loading page...")
driver.get(url)
time.sleep(8)  # JSロード待ち（長めに）

html = driver.page_source
driver.quit()

print("HTML length:", len(html))
soup = BeautifulSoup(html, "html.parser")

data = []

def add(category, spec, value):
    print(f"[ADD] {category} | {spec} | {value}")
    data.append([category.strip(), spec.strip(), value.strip()])

# -------------------------
# 1. カテゴリ（h2/h3）検出
# -------------------------
print("\n=== Detecting categories (h2/h3) ===")
sections = soup.find_all(["section", "div"])
for sec in sections:
    h = sec.find(["h2", "h3"])
    if h:
        print("Category found:", h.get_text(strip=True))

# -------------------------
# 2. table 構造の検出
# -------------------------
print("\n=== Detecting <table> ===")
tables = soup.find_all("table")
print("Number of tables found:", len(tables))

for t in tables:
    print("Table HTML snippet:", str(t)[:200], "...\n")

# -------------------------
# 3. dl 構造の検出
# -------------------------
print("\n=== Detecting <dl> ===")
dls = soup.find_all("dl")
print("Number of dl found:", len(dls))

for d in dls:
    print("DL snippet:", str(d)[:200], "...\n")

# -------------------------
# 4. spec-key / spec-value の検出
# -------------------------
print("\n=== Detecting .spec-key / .spec-value ===")
keys = soup.select(".spec-key")
values = soup.select(".spec-value")
print("Keys:", len(keys), "Values:", len(values))

# -------------------------
# 5. spec-item の検出
# -------------------------
print("\n=== Detecting .spec-item ===")
items = soup.select(".spec-item")
print("spec-item count:", len(items))

# -------------------------
# 6. 実際の抽出処理
# -------------------------
print("\n=== Extracting data ===")

for sec in sections:
    h = sec.find(["h2", "h3"])
    if not h:
        continue
    category = h.get_text(strip=True)

    # table
    for tr in sec.select("table tr"):
        th = tr.find("th")
        td = tr.find("td")
        if th and td:
            add(category, th.get_text(), td.get_text(" ", strip=True))

    # dl
    for dt in sec.find_all("dt"):
        dd = dt.find_next("dd")
        if dd:
            add(category, dt.get_text(), dd.get_text(" ", strip=True))

    # spec-key/value
    keys = sec.select(".spec-key")
    values = sec.select(".spec-value")
    for k, v in zip(keys, values):
        add(category, k.get_text(), v.get_text(" ", strip=True))

    # spec-item
    for item in sec.select(".spec-item"):
        label = item.select_one(".label")
        value = item.select_one(".value")
        if label and value:
            add(category, label.get_text(), value.get_text(" ", strip=True))

df = pd.DataFrame(data, columns=["Category", "Spec", "Value"])
df.to_excel("hp_specs_debug.xlsx", index=False)

print("\nDone.")
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

url = "https://www.brother.ca/en/p/hll2460dw"
res = requests.get(url)
soup = BeautifulSoup(res.text, "html.parser")

rows = []

# --- 1) テーブル形式の仕様を抽出 ---
for table in soup.find_all("table"):
    # カテゴリ名（テーブルの直前の見出しを取得）
    heading = table.find_previous(["h2", "h3", "h4"])
    category = heading.get_text(strip=True) if heading else "Unknown"

    for tr in table.find_all("tr"):
        cells = tr.find_all(["td", "th"])
        if len(cells) >= 2:
            spec = cells[0].get_text(" ", strip=True)
            value = cells[1].get_text(" ", strip=True)
            rows.append([category, spec, value])

# --- 2) <dl>（定義リスト）形式の仕様を抽出 ---
for dl in soup.find_all("dl"):
    heading = dl.find_previous(["h2", "h3", "h4"])
    category = heading.get_text(strip=True) if heading else "Unknown"

    dts = dl.find_all("dt")
    dds = dl.find_all("dd")

    for dt, dd in zip(dts, dds):
        spec = dt.get_text(" ", strip=True)
        value = dd.get_text(" ", strip=True)
        rows.append([category, spec, value])

# --- 3) その他の仕様ブロック（div 形式） ---
spec_blocks = soup.select("div.specifications, div.product-specifications")
for block in spec_blocks:
    heading = block.find_previous(["h2", "h3", "h4"])
    category = heading.get_text(strip=True) if heading else "Unknown"

    items = block.find_all("div")
    for item in items:
        parts = item.get_text(" ", strip=True).split(":")
        if len(parts) == 2:
            spec, value = parts
            rows.append([category, spec.strip(), value.strip()])

# DataFrame にまとめる
df = pd.DataFrame(rows, columns=["Category", "Spec", "Value"])
df
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

url = "https://jp.ext.hp.com/printers/business-printers/laserjet/enterprise_5700dn/"
res = requests.get(url)
soup = BeautifulSoup(res.text, "html.parser")

rows = []

# --- 1) 仕様テーブル（<table>）をすべて抽出 ---
for table in soup.find_all("table"):
    # テーブルの直前の見出し（カテゴリ名）を取得
    heading = table.find_previous(["h2", "h3", "h4"])
    category = heading.get_text(strip=True) if heading else "カテゴリ不明"

    for tr in table.find_all("tr"):
        cells = tr.find_all(["td", "th"])
        if len(cells) >= 2:
            spec = cells[0].get_text(" ", strip=True)
            value = cells[1].get_text(" ", strip=True)
            rows.append([category, spec, value])

# --- 2) 定義リスト（<dl><dt><dd>）形式も抽出（HP で使われることがある） ---
for dl in soup.find_all("dl"):
    heading = dl.find_previous(["h2", "h3", "h4"])
    category = heading.get_text(strip=True) if heading else "カテゴリ不明"

    dts = dl.find_all("dt")
    dds = dl.find_all("dd")

    for dt, dd in zip(dts, dds):
        spec = dt.get_text(" ", strip=True)
        value = dd.get_text(" ", strip=True)
        rows.append([category, spec, value])

# --- 3) その他の仕様ブロック（div 形式）も拾う ---
for block in soup.select("div.spec, div.specification, div.hp-spec"):
    heading = block.find_previous(["h2", "h3", "h4"])
    category = heading.get_text(strip=True) if heading else "カテゴリ不明"

    items = block.find_all("div")
    for item in items:
        text = item.get_text(" ", strip=True)
        if ":" in text:
            spec, value = text.split(":", 1)
            rows.append([category, spec.strip(), value.strip()])

# --- DataFrame にまとめる ---
df = pd.DataFrame(rows, columns=["Category", "Spec", "Value"])
df
```
